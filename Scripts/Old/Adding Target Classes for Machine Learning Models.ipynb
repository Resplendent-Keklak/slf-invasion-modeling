```python
def assign_targets(gdf: gpd.GeoDataFrame = general_county_training, specificity: str = 'county', index_cutoff: int = 3108, purpose: str = 'general') -> gpd.GeoDataFrame:
    def check_specificity(level):
        if level != 'county':
            return 'state'

    specificity = check_specificity(specificity)
    
    targets_list = []
    for i in range(0, len(gdf)):
        # try:
        #     if gdf.at[i, 'year'] == 2014 and gdf.at[i, 'county'] != 'Berks' and gdf.at[i, 'state'] != 'Pennsylvania':
        #         targets_list = targets_list + [0]
                
        # except KeyError:
        #     print('County or year column not found')

        # else:
        #     if gdf.at[i, 'year'] == 2014 and gdf.at[i, 'state'] != 'Pennsylvania':
        #         targets_list = targets_list + [0]

        # finally:
        #     pass

        # if gdf.at[i, 'year'] == 2014:
        #     targets_list = targets_list + [1] # might have actually been more but I'm being conserative here

        if specificity == 'county' and gdf.at[i, 'year'] == 2014 and gdf.at[i, 'county'] == 'Berks' and gdf.at[i, 'state'] == 'Pennsylvania':
            targets_list = targets_list + [1] # might have actually been more but I'm being conserative here

        elif specificity == 'state' and gdf.at[i, 'year'] == 2014 and gdf.at[i, 'state'] == 'Pennsylvania':
            targets_list = targets_list + [1]

        elif gdf.at[i, 'year'] == 2014:
            targets_list = targets_list + [0]
            
        elif purpose == 'general' or (purpose != 'food' and purpose != 'fiber' and purpose != 'ornamental'):
            if specificity == 'county':
                current_county = gdf.at[i, 'county']
            
            current_year = gdf.at[i, 'year']
            current_state = gdf.at[i, 'state']
            current_density = gdf.at[i, 'slfdensity']
            previous_density = gdf.at[i, 'y_slfdnsty']
            # previous_data_point = gdf.loc[gdf['year'] == current_year - 1]
            # previous_data_point = previous_data_point.loc[gdf['state'] == current_state] # I have to make subset using three different assignments,
            # previous_data_point = previous_data_point.loc[gdf['county'] == current_county] # so please bear with me
            
            # if len(previous_data_point) == 0:
            #     print('Not going to be able to use comparisons in risk assessment. Will assume the previous year\'s density is 0 SLF/sq km')
            #     previous_density = 0
                
            # elif len(previous_data_point) > 1:
            #     print('Multiple values may determine risk assessment target values.')
            #     previous_density = previous_data_point['slfdensity'].sum() / len(previous_data_point) # take unweighted average
                
            # else:
            #     previous_density = previous_data_point.at[previous_data_point.index.to_list()[0], 'slfdensity']

            # Find the previous target value
            
            if i < index_cutoff:
                previous_target = 0

            else:
                previous_target = targets_list[i - index_cutoff]

            # Make current target's assignment; risk levels = relative difference
            
            if previous_target == 0 and current_density == 0:
                targets_list = targets_list + [0]
            
            elif current_density < 1e2 or current_density == 1e2 or current_density - previous_density < 1e2 or current_density - previous_density == 1e2:
                targets_list = targets_list + [1]

            elif current_density - previous_density < 1e3 or current_density - previous_density == 1e3:
                targets_list = targets_list + [2]

            elif current_density - previous_density < 1e4 or current_density - previous_density == 1e4:
                targets_list = targets_list + [3]

            else:
                targets_list = targets_list + [4] # verify this is not an anomaly
            
        elif purpose == 'food':
            pass
            
        elif purpose == 'timber and paper' or purpose == 'fiber':
            pass
            
        elif purpose == 'ornamental':
            current_county = gdf.at[i, 'county']
            county_ornamental_count = gdf.at[i, 'odiversity']
            county_ornamental_density = gdf.at[i, 'oabundance'] / gdf.at[i, 'land_area']
            county_non_ornamental_host_count = gdf.at[i, 'non_ornamental_host_species_diversity']
            county_non_ornamental_density = gdf.at[i, 'est_non_ornamental_abundance'] / gdf.at[i, 'land area']
            current_density = gdf.at[i, 'estimated_slf_density']
            lanternflies_per_host = current_density / (county_ornamental_density + county_non_ornamental_density)
            lanternflies_per_ornamental = lanternflies_per_host * county_ornamental_density / (county_ornamental_density + county_non_ornamental_density) 
                # assuming equal feeding
            if lanternflies_per_host < 100 and lanternflies_per_ornamental < lanternflies_per_host:
                targets_list = targets_list + [0] # this is not enough to cause signfiicant damage to ornamentals

            elif lanternflies_per_ornamental < 500 and (lanternflies_per_ornamental <= lanternflies_per_host or county_ornamental_count >= county_non_ornamental_host_count):
                targets_list = targets_list + [1] # this may be enough to cause statistically signfiicant damage to some ornamentals

            else:
                targets_list = targets_list + [2] # this may enough to cause statistically signfiicant damage to a majority of ornamentals
            
        # else:
            # print('Purpose not defined. Choosing general spread risk...')
            # repeat code snippet from general condition here
            
    return targets_list
```

```python
# 2025 example

possibly_2025 = general_county_training.iloc[3108 * 10 : 3108 * 11]
possibly_2025 = possibly_2025.reset_index()
changes = []
chameleon = []
for i in range(0, len(possibly_2025)):
    this_target = possibly_2025.at[i, 'target']
    this_state = possibly_2025.at[i, 'state']
    if this_target == 4 and this_state not in ['Pennsylvania', 'New Jersey', 'New York', 'Maryland', 'Delaware', 'Massachusetts', 'Connecticut',
                                               'Rhode Island', 'Vermont']:
        changes = changes + [1]
        chameleon = chameleon + ['#80FF80']
        
    elif this_target == 4:
        changes = changes + [2]
        chameleon = chameleon + ['#FFFFC0']
    elif this_target == 0 and this_state in ['Pennsylvania', 'New Jersey', 'New York', 'Maryland', 'Delaware', 'Massachusetts', 'Connecticut',
                                             'Rhode Island', 'Vermont']:
        changes = changes + [2]
        chameleon = chameleon + ['#FFFFC0']
    else:
        changes = changes + [possibly_2025.at[i, 'target']]
        chameleon = chameleon + [possibly_2025.at[i, 'color']]

possibly_2025['target'] = changes
possibly_2025['color'] = chameleon

possibly_2025.plot(color = possibly_2025['color'])
```
