```python
# Read columns
# for i in range(0, len(years_list)):
#     print(years_list[i].columns)

# Create a minor function that converts LCD timestamps to datetimes so that they can be easily parsed

def to_time(column):
    list_times = []
    for timestamp in range(0, len(column)):
        new_timestamp = str(column[timestamp])
        new_timestamp = datetime.datetime.strptime(new_timestamp, '%Y-%m-%dT%H:%M:%S')
        list_times = list_times + [new_timestamp]
    return list_times
```
