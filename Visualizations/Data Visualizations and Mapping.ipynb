## New Maps Using Crimson Scale

```python
# New reds-only map plus categorical legend

def map(gdf: gpd.GeoDataFrame, level: str, purpose: str, year: int, colorcolumn: str, indices: list):
    fig, ax = plt.subplots(1, 1)
    ax.set_xlabel('Longitude')
    ax.set_ylabel('Latitude')

    level = level.lower()
    if level != 'county' and level != 'state':
        level = 'state'

    purpose = purpose.lower()
    if purpose != 'general' and purpose != 'food' and purpose != 'fiber' and purpose != 'ornamental':
        titleformat = 'Population Density of Spotted Lanternflies (per square kilometer) in ' + str(year)
        plt.title(titleformat)
        gdf.iloc[indices].plot(column = colorcolumn,
                               ax = ax,
                               cax = make_axes_locatable(ax).append_axes('bottom', size = '5%', pad = 0.5),
                               cmap = 'Reds',
                               legend = True,
                               legend_kwds={'label': 'Invasion Density', 'orientation': 'horizontal'}
                              )
        
    else:
        if purpose == 'general':
            purpose = 'Invasion Risks'
            categories = [0, 1, 2, 3, 4]

        elif purpose == 'food':
            purpose = 'Risks to Plants Grown for Food Production'
            categories = [0, 1, 2, 3]

        elif purpose == 'fiber':
            purpose = 'Risk to Plants Grown for Timber, Paper, or Fiber Production'
            categories = [0, 1, 2, 3]

        else:
            purpose = 'Risk to Plants Grown for Ornamental Use'
            categories = [0, 1, 2]
        
        titleformat = level[0].upper() + level[1:] + '-Based Spotted Lanternfly ' + purpose + ' in ' + str(year)
        plt.title(titleformat)

        gdf.iloc[indices].plot(column = colorcolumn,
                               ax = ax,
                               cax = make_axes_locatable(ax),
                               categorical = True,
                               categories = categories,
                               vmin = 0,
                               vmax = max(categories),
                               cmap = 'Reds',
                               legend = True
                              )
    return None
```

## Old Maps Using Green-to-Yellow-to-Red Scale

```python
# Original green-to-red map

def old_map(gdf: gpd.GeoDataFrame, k: int, indices: list):
    if k != 4 and k != 3:
        palette = {0: '#008000', 1: '#80FF80', 2: '#FFFFC0', 3: '#FFC000', 4: '#800000'}
    
    elif k == 4:
        palette = {0: '#008000', 1: '#FFFFC0', 2: '#FFC000', 3: '#800000'}
    
    else:
        palette = {0: '#008000', 1: '#FFFFC0', 2: '#800000'}

    # Probably check here if the coloring column's contents are of the set 0, 1, 2, 3, 4 and such

    # Probably also check if the color column already exists
    gdf.drop('color', axis = 1)
    
    gdf['color'] = [palette[gdf.at[i, 'target']] for i in range(0, len(gdf))]
    
    gdf.iloc[indices].plot(color = gdf.iloc[indices]['color'])
    return 0
```
